```python
import json  
import pickle
import time  
from pathlib import Path  

import numpy as np  
import pandas as pd 
import torch  
from sklearn.metrics import classification_report, confusion_matrix 
from sklearn.model_selection import train_test_split  
from transformers import pipeline, set_seed  

SEED = 42  
set_seed(SEED)
```

```python
SPLITS_DIR = Path("../../data/splits")  
ARTIFACTS_DIR = Path("../../data/artifacts")  
RESULTS_DIR = Path("../../../results/zeroshot")
RESULTS_DIR.mkdir(exist_ok=True, parents=True)

test_df = pd.read_csv(SPLITS_DIR / "test.csv") 

with open(ARTIFACTS_DIR / "label_encoder.pkl", "rb") as f:
    label_encoder = pickle.load(f)  

LABEL_ORDER = list(label_encoder.classes_) 
NUM_CLASSES = len(LABEL_ORDER)

print("Test set size:", len(test_df))
print("Classes:", LABEL_ORDER)
```

```python
MODEL_CHECKPOINT = "facebook/bart-large-mnli" 

DESCRIPTIVE_LABEL_MAP = {
    "Business": "business and finance",
    "Entertainment": "entertainment, celebrities, and movies",
    "Health": "health, wellness, and medicine",
    "Politics": "politics and government",
    "Science": "science and research",
    "Sports": "sports",
    "Technology": "technology",
    "World": "world news and international affairs",
}
CANDIDATE_LABELS = [DESCRIPTIVE_LABEL_MAP[c] for c in LABEL_ORDER]  
PHRASE_TO_LABEL = {v: k for k, v in DESCRIPTIVE_LABEL_MAP.items()}  

HYPOTHESIS_TEMPLATE = "This news article is about {}."

BATCH_SIZE = 32  
```

```python
EVAL_ON_FULL_TEST = False  
SUBSAMPLE_SIZE = 2500  

if EVAL_ON_FULL_TEST:
    eval_df = test_df.reset_index(drop=True)
    sampling_strategy = "full_test_set"
else:
    eval_df, _ = train_test_split(
        test_df,
        train_size=SUBSAMPLE_SIZE,
        stratify=test_df["category"],  
        random_state=SEED,
    )
    eval_df = eval_df.reset_index(drop=True)
    sampling_strategy = "stratified_subsample"

print(f"Evaluating on {len(eval_df)} / {len(test_df)} test rows ({sampling_strategy})")
print(eval_df["category"].value_counts())
```

```python
device = 0 if torch.cuda.is_available() else -1  
classifier = pipeline(
    "zero-shot-classification",
    model=MODEL_CHECKPOINT,
    device=device,
)
```

```python
texts = eval_df["text"].tolist()  

inf_start = time.time()  
raw_results = classifier(
    texts,
    candidate_labels=CANDIDATE_LABELS,  
    hypothesis_template=HYPOTHESIS_TEMPLATE,  
    multi_label=False,  
    batch_size=BATCH_SIZE,
)
inf_time_sec = time.time() - inf_start  

y_pred = [PHRASE_TO_LABEL[r["labels"][0]] for r in raw_results]  
y_test = eval_df["category"].tolist()  

print(f"Inference time: {inf_time_sec:.2f}s total, {inf_time_sec / len(eval_df) * 1000:.3f} ms/sample")
```

```python
report_dict = classification_report(
    y_test, y_pred, labels=LABEL_ORDER, output_dict=True, zero_division=0  
)
report_text = classification_report(y_test, y_pred, labels=LABEL_ORDER, zero_division=0)  
print(report_text)

cm = confusion_matrix(y_test, y_pred, labels=LABEL_ORDER)  
test_accuracy = report_dict["accuracy"]
print(f"Test accuracy: {test_accuracy:.4f}")
```

```python
import matplotlib.pyplot as plt  
import seaborn as sns 

plt.figure(figsize=(8, 7))  # square-ish figure sized for an 8-class matrix
sns.heatmap(cm, annot=True, fmt="d", cmap="Blues", xticklabels=LABEL_ORDER, yticklabels=LABEL_ORDER)
plt.xlabel("Predicted")
plt.ylabel("True")
plt.title("Zero-Shot (tuned: descriptive labels + custom template) - Test Confusion Matrix")
plt.tight_layout()
plt.savefig(RESULTS_DIR / "ZeroShot_b_confusion_matrix.png", dpi=150)
plt.show()
```

```python
metrics = {
    "model_family": "zeroshot",  
    "model_name": MODEL_CHECKPOINT,
    "variant": "tuned",
    "run_timestamp": pd.Timestamp.now("UTC").isoformat(),  
    "label_order": LABEL_ORDER,  
    "config": {
        "model_checkpoint": MODEL_CHECKPOINT,
        "hypothesis_template": HYPOTHESIS_TEMPLATE,
        "candidate_labels": CANDIDATE_LABELS,
        "descriptive_label_map": DESCRIPTIVE_LABEL_MAP,
        "multi_label": False,
        "batch_size": BATCH_SIZE,
        "full_test_size": len(test_df),  
        "sampling_strategy": sampling_strategy,  
    },
    "train": {
        "n_train": 0,  
        "n_val": 0,  
        "train_time_sec": 0.0,  
        "epochs_trained": None,
        "history": {},
    },
    "test": {
        "n_test": len(eval_df),  
        "accuracy": float(test_accuracy),
        "macro_precision": float(report_dict["macro avg"]["precision"]),
        "macro_recall": float(report_dict["macro avg"]["recall"]),
        "macro_f1": float(report_dict["macro avg"]["f1-score"]),
        "weighted_f1": float(report_dict["weighted avg"]["f1-score"]),
        "per_class": {
            cls: {
                "precision": float(report_dict[cls]["precision"]),
                "recall": float(report_dict[cls]["recall"]),
                "f1": float(report_dict[cls]["f1-score"]),
                "support": int(report_dict[cls]["support"]),
            }
            for cls in LABEL_ORDER
        },
        "confusion_matrix": cm.tolist(),
        "inference_time_sec_total": inf_time_sec,
        "inference_time_ms_per_sample": inf_time_sec / len(eval_df) * 1000,
    },
    "environment": {
        "framework": "pytorch+transformers",
        "framework_version": torch.__version__,
        "device": "GPU" if torch.cuda.is_available() else "CPU",
        "seed": SEED,
    },
}

with open(RESULTS_DIR / "ZeroShot_b_metrics.json", "w") as f:
    json.dump(metrics, f, indent=2)  

print("Saved:", RESULTS_DIR / "ZeroShot_b_metrics.json")
```
